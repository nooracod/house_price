```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib


data = {
    'area': [50, 80, 120, 150, 200, 65, 95],
    'floor': [1, 3, 2, 5, 1, 4, 2],
    'year_built': [1395, 1390, 1400, 1385, 1402, 1392, 1398],
    'parking': [0, 1, 1, 1, 2, 0, 1],
    'warehouse': [0, 1, 1, 0, 1, 0, 1],
    'price': [1500, 3000, 5000, 5500, 8000, 2100, 3800]
}

df = pd.DataFrame(data)


X = df[['area', 'floor', 'year_built', 'parking', 'warehouse']]
y = df['price']


model = LinearRegression()
model.fit(X, y)


joblib.dump(model, 'model.pkl')
print("مدل با موفقیت در گوگل کولب ساخته شد و در فایل ذخیره شد!")
```

```python
import numpy as np
import joblib


loaded_model = joblib.load('model.pkl')

sample_house = np.array([[100, 2, 1395, 1, 1]])

predicted_price = loaded_model.predict(sample_house)[0]
print(f"قیمت پیش‌بینی شده برای این خانه: {round(predicted_price, 2)} میلیون تومان")
```

```python

!pip install fastapi uvicorn scikit-learn pandas numpy joblib nest-asyncio pyngrok > /dev/null 2>&1

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import joblib
import nest_asyncio
import threading
import uvicorn
from fastapi import FastAPI, Form
from fastapi.responses import HTMLResponse
from pyngrok import ngrok


data = {
    'area': [50, 80, 120, 150, 200, 65, 95],
    'floor': [1, 3, 2, 5, 1, 4, 2],
    'year_built': [1395, 1390, 1400, 1385, 1402, 1392, 1398],
    'parking': [0, 1, 1, 1, 2, 0, 1],
    'warehouse': [0, 1, 1, 0, 1, 0, 1],
    'price': [1500, 3000, 5000, 5500, 8000, 2100, 3800]
}
df = pd.DataFrame(data)
X = df[['area', 'floor', 'year_built', 'parking', 'warehouse']]
y = df['price']

model = LinearRegression()
model.fit(X, y)


app = FastAPI()

html_content = """
<!DOCTYPE html>
<html lang="fa" dir="rtl">
<head>
    <meta charset="UTF-8">
    <title>پیش‌بینی قیمت مسکن</title>
    <style>
        body { font-family: Tahoma, sans-serif; background: #f4f4f9; padding: 40px; }
        .box { max-width: 400px; background: white; padding: 20px; border-radius: 8px; margin: auto; box-shadow: 0 0 10px rgba(0,0,0,0.1); }
        input, select { width: 100%; padding: 8px; margin: 8px 0; box-sizing: border-box; }
        button { width: 100%; background: #28a745; color: white; padding: 10px; border: none; border-radius: 4px; cursor: pointer; font-size: 16px; }
        button:hover { background: #218838; }
        .result { margin-top: 15px; padding: 12px; background: #d4edda; color: #155724; text-align: center; border-radius: 4px; font-weight: bold; }
    </style>
</head>
<body>
<div class="box">
    <h2>پیش‌بینی قیمت خانه</h2>
    <form method="post">
        <label>متراژ (متر مربع):</label>
        <input type="number" step="any" name="area" required>

        <label>طبقه:</label>
        <input type="number" name="floor" required>

        <label>سال ساخت:</label>
        <input type="number" name="year_built" required>

        <label>پارکینگ:</label>
        <select name="parking">
            <option value="1">دارد</option>
            <option value="0">ندارد</option>
        </select>

        <label>انباری:</label>
        <select name="warehouse">
            <option value="1">دارد</option>
            <option value="0">ندارد</option>
        </select>

        <button type="submit">محاسبه قیمت</button>
    </form>
</div>
</body>
</html>
"""

@app.get("/", response_class=HTMLResponse)
def home():
    return html_content

@app.post("/", response_class=HTMLResponse)
def predict(
    area: float = Form(...),
    floor: int = Form(...),
    year_built: int = Form(...),
    parking: int = Form(...),
    warehouse: int = Form(...)
):
    user_data = np.array([[area, floor, year_built, parking, warehouse]])
    predicted_price = round(model.predict(user_data)[0], 2)

    result_html = html_content + f"""
    <script>
        window.addEventListener('DOMContentLoaded', (event) => {{
            let div = document.createElement('div');
            div.className = 'result';
            div.innerHTML = 'قیمت تقریبی خانه: {predicted_price} میلیون تومان';
            document.querySelector('.box').appendChild(div);
        }});
    </script>
    """
    return result_html


def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

server_thread = threading.Thread(target=run_server)
server_thread.start()

try:
    ngrok.kill()
    public_url = ngrok.connect(8000)
    print("✨ سایت شما با موفقیت آماده شد! روی لینک زیر کلیک کنید:")
    print(public_url)
except Exception as e:
    print("خطا در ساخت لینک ngrok:", e)


```
